```python
def calc_grade(marks):
  if marks > 90:
    grade = "A"
  elif marks > 75:
    grade = "B"
  elif marks > 60:
    grade = "C"
  else:
    grade = "D"
  return grade
```

```python
for student in students:
  print(student["marks"], calc_grade(student["marks"]))
```
